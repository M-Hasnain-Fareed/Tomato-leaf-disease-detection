# Project: Tomato Leaf Disease Detection using YOLO26n

## Description: Fine-tuning a custom YOLO26 nano model for agricultural research.


```python
import os
from pathlib import Path
import torch
from ultralytics import YOLO

# Display environment & hardware specifications for reproducibility in research
print(f"PyTorch Version: {torch.__version__}")
print(f"CUDA Available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"GPU Device: {torch.cuda.get_device_name(0)}")
```

# Configure dataset paths and verify structure

```python
DATASET_NAME = "Tomato Leaf Disease.v63i.yolo26"
BASE_DIR = Path.cwd()
DATA_YAML_PATH = BASE_DIR / DATASET_NAME / "data.yaml"

if not DATA_YAML_PATH.exists():
    raise FileNotFoundError(
        f"Configuration file not found at {DATA_YAML_PATH}. Please verify your directory structure."
    )

print(f"Dataset configuration successfully located at: {DATA_YAML_PATH}")
```

# Initialize the pretrained baseline architecture (Transfer Learning)

```python
MODEL_NAME = "yolo26n.pt"

print(f"Loading pretrained base model: {MODEL_NAME}")
model = YOLO("yolo26n.pt")
```

# Execute fine-tuning process with research-grade hyperparameters


```python
EPOCHS = 50
EPOCHS = 50
IMAGE_SIZE = 640
BATCH_SIZE = 4
PROJECT_NAME = "tomato_disease_research"
RUN_NAME = "yolo26n_finetuned"

print(f"Starting training for {EPOCHS} epochs with batch size {BATCH_SIZE}...")

training_results = model.train(
    data=str(DATA_YAML_PATH),
    epochs=EPOCHS,
    imgsz=IMAGE_SIZE,
    batch=BATCH_SIZE,
    workers=0,
    project=PROJECT_NAME,
    name=RUN_NAME,
    device=0,
    save=True,
    plots=True 
)

print("Training sequence completed successfully.")
```

# Validate the best-performing weights on the validation split to extract key metrics

```python
best_weights_path = BASE_DIR / PROJECT_NAME / RUN_NAME / "weights" / "best.pt"

print(f"Loading best weights from: {best_weights_path}")
validated_model = YOLO(str(best_weights_path))

print("Running validation evaluation...")
metrics = validated_model.val(data=str(DATA_YAML_PATH))

# Output core quantitative metrics for documentation or manuscript integration
print("\n--- Quantitative Evaluation Results ---")
print(f"mAP50-95: {metrics.box.map:.4f}")
print(f"mAP50:     {metrics.box.map50:.4f}")
print(f"Precision: {metrics.box.p.mean():.4f}")
print(f"Recall:    {metrics.box.r.mean():.4f}")
```

# Inference: Run the fine-tuned model on a sample test image

```python
import glob
from IPython.display import Image, display

# Locate a test image from your test split directory
test_image_paths = glob.glob(str(BASE_DIR / DATASET_NAME / "test" / "images" / "*.jpg")) + \
                   glob.glob(str(BASE_DIR / DATASET_NAME / "test" / "images" / "*.png"))

if test_image_paths:
    sample_img = test_image_paths[0]
    print(f"Running inference on test sample: {sample_img}")
    
    # Predict using the saved best weights from your training run
    predict_results = validated_model.predict(
        source=sample_img,
        conf=0.35,  # Confidence threshold filter
        save=True   # Automatically saves the output image with bounding boxes
    )
    print("Inference complete. Results saved under the runs directory.")
else:
    print("No test images found in the specified path.")
```

# Displaying result graphs

```python
plots_dir = BASE_DIR / PROJECT_NAME / RUN_NAME

print("--- Displaying Core Research Plots ---")

# 1. Training metrics evolution plot
results_graph = plots_dir / "results.png"
if results_graph.exists():
    print("\nTraining Metrics Progression (results.png):")
    display(Image(filename=str(results_graph), width=800))

# 2. Confusion Matrix
confusion_matrix_graph = plots_dir / "confusion_matrix.png"
if confusion_matrix_graph.exists():
    print("\nConfusion Matrix (confusion_matrix.png):")
    display(Image(filename=str(confusion_matrix_graph), width=800))

# 3. Precision-Recall Curve
pr_curve_graph = plots_dir / "BoxPR_curve.png"
if pr_curve_graph.exists():
    print("\nPrecision-Recall Curve (BoxPR_curve.png):")
    display(Image(filename=str(pr_curve_graph), width=800))
```
